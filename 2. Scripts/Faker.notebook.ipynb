```python

# Librerías para el Proyecto Integrador FleetLogix

import pandas as pd
import numpy as np
from faker import Faker
import psycopg2
import random
from datetime import datetime, timedelta

# Inicializar Faker
fake = Faker("es_AR")

print("Librerías importadas correctamente")
```

```python

# Generación de vehículos de prueba

tipos_vehiculos = [
    "Camión Grande",
    "Camión Mediano",
    "Van",
    "Motocicleta"
]

vehiculos = []

for i in range(5):
    vehiculo = {
        "license_plate": f"FLX-{i+1:04d}",
        "vehicle_type": random.choice(tipos_vehiculos),
        "capacity_kg": 1000,
        "fuel_type": "Diesel",
        "acquisition_date": fake.date_between(
            start_date="-5y",
            end_date="-2y"
        ),
        "status": "active"
    }

    vehiculos.append(vehiculo)

df_vehicles = pd.DataFrame(vehiculos)

display(df_vehicles)
```
